# 🔬 FD001 Analysis

This notebook reproduces the engine-level benchmark analysis from the NASA C-MAPSS FD001 test RUL file. The supplied RUL values are ground-truth labels for evaluation.

```python
from pathlib import Path
import pandas as pd
from src.aerospace_analytics import build_evaluation_table, summarize

path = Path('data/raw/RUL_FD001.txt')
rul = pd.read_csv(path, header=None, names=['true_rul_cycles'])
rul['engine_id'] = range(1, len(rul) + 1)
result = build_evaluation_table(rul[['engine_id', 'true_rul_cycles']])
metrics = summarize(result)
metrics
```
